## 1. Phonology

Phonology deals with the sound systems of languages. In a computational setting, we might represent words by their phonetic or phonemic transcriptions.

Below, we show a simple *demonstration* using a small lookup table of words to their *approximate* phonemic representations. (In real analysis, you would use a library like [CMU Pronouncing Dictionary](http://www.speech.cs.cmu.edu/cgi-bin/cmudict) or other tools.)

```python
# Simple dictionary mapping words to approximate phonetic transcriptions
phonetic_dict = {
    "cat": "kæt",
    "dog": "dɔg",
    "computer": "kəmˈpjuːtər",
    "python": "ˈpaɪθɑn"
}

words_to_test = ["cat", "dog", "python", "computer"]

for w in words_to_test:
    phonetic = phonetic_dict.get(w, "[Phonetic not found]")
    print(f"Word: {w}\tPhonetic: {phonetic}")
```

CMU Pronouncing Dictionary: A Python library containing phonetic transcriptions for thousands of English words in ARPAbet format.

```python
import nltk
from nltk.corpus import cmudict

nltk.download('cmudict')
pronunciations = cmudict.dict()

word = "computer"
phonetic = pronunciations.get(word.lower(), "Phonetic not found")
print(phonetic)
```

## 2. Morphology

Morphology analyzes how words are formed from smaller units (morphemes). In English, we often look at roots, prefixes, and suffixes. Below we use **spaCy** to retrieve morphological features of each token in a sentence.

Lemma
- The lemma is the base or dictionary form of a word. For example:

-- cats → cat (the singular form)

-- playing → play (the infinitive form)

-- are → be (the base verb)

Lemmatization is useful for standardizing words and identifying their root meaning.

```python
# Example sentence to analyze
sentence = "The cats are playing happily in the backyard."

# Process sentence with spaCy
doc = nlp(sentence)

print(f"Analyzing Morphology for: '{sentence}'\n")
for token in doc:
    # text, lemma, morphological features
    print(
        f"Token: {token.text}\tLemma: {token.lemma_}\tMORPH: {token.morph}\tPOS: {token.pos_}"
    )
```

MORPH (Morphological Features)

The MORPH column specifies grammatical features of the token, such as tense, number, aspect, mood, or case. Here's a breakdown of the MORPH features provided:

Determiners (DET)

The:

Definite=Def → Indicates it is a definite article.

PronType=Art → Specifies that it is an article.

Nouns (NOUN)

cats:

Number=Plur → Indicates that the noun is plural.

backyard:

Number=Sing → Indicates that the noun is singular.

Auxiliary Verb (AUX)

are:

Mood=Ind → The verb is in the indicative mood (used for factual statements).
Tense=Pres → The verb is in the present tense.

VerbForm=Fin → The verb is finite, meaning it shows tense and agrees with the subject.

Main Verb (VERB)

playing:

Aspect=Prog → Indicates progressive aspect (ongoing action).

Tense=Pres → The verb is in the present tense.

VerbForm=Part → It is a participle form (used in a compound tense).

Adverb (ADV)

happily: No morphological features are provided for adverbs in this analysis.

Preposition (ADP)

in: No morphological features are provided for prepositions in this analysis.

Punctuation (PUNCT)
.:
PunctType=Peri → Indicates that it is a period (full stop).

POS (Part of Speech)

The POS column categorizes each token into its grammatical role:

DET (Determiner): Words like "the" that introduce nouns.

NOUN (Noun): Words that represent people, places, or things (e.g., cats, backyard).

AUX (Auxiliary): Helping verbs that support the main verb (e.g., are).

VERB (Verb): Words that describe actions or states (e.g., playing).

ADV (Adverb): Words that modify verbs, adjectives, or other adverbs (e.g., happily).

ADP (Adposition): Prepositions or postpositions (e.g., in).

PUNCT (Punctuation): Symbols like periods, commas, or quotation marks (e.g., .).

## 3. Syntax

Syntax is about how words combine to form phrases and sentences. Below we show **spaCy**’s dependency parse of a sentence.

```python
sentence_syntax = "I gave my friend a birthday present."  # Feel free to change
doc_syntax = nlp(sentence_syntax)

print(f"Syntactic Dependencies for: '{sentence_syntax}'\n")
for token in doc_syntax:
    print(
        f"Token: {token.text}\tPOS: {token.pos_}\tHead: {token.head.text}\tDep: {token.dep_}"
    )
```

## 4. Lexicon

A **lexicon** is essentially a dictionary of a language. We can explore lexical information (like parts of speech, definitions, synonyms, etc.) using **NLTK**’s WordNet. Below, we retrieve synonyms for a given word.

```python
from nltk.corpus import wordnet

word = "play"
synsets = wordnet.synsets(word)

print(f"Synsets for '{word}':\n")
for s in synsets:
    print("Synset:", s.name())
    print("Definition:", s.definition())
    print("Examples:", s.examples())
    print("---------------------")
```

## 5. Semantics

Semantics deals with the meaning of words and sentences. One simple way to explore semantics is by looking at **semantic similarity** between words using pretrained embeddings (spaCy has built-in vectors for `en_core_web_md` or `en_core_web_lg`, but for `en_core_web_sm` the vectors might be limited).

Below, we’ll do a basic similarity check. If you have a larger spaCy model installed (e.g., `en_core_web_md`), replace `'en_core_web_sm'` above with `'en_core_web_md'`.

```python
tokens = nlp("cat dog apple")

for t1 in tokens:
    for t2 in tokens:
        if t1 != t2:
            print(f"Similarity({t1.text}, {t2.text}) = {t1.similarity(t2):.4f}")
```

## 6. Pragmatics

Pragmatics involves how context influences the interpretation of meaning. It’s harder to demonstrate pragmatics computationally in a simple script, because it often requires **context** or **world knowledge**.

However, we can illustrate a tiny example by reusing spaCy’s named entity recognition (NER) to see how it identifies references to entities in context. The same word can refer to different entities depending on context. Modify the text to see how spaCy’s NER changes with context.

```python
context_sentence = "Apple is looking at buying U.K. startup for $1 billion."
doc_context = nlp(context_sentence)

print(f"Context: '{context_sentence}'\n")
for ent in doc_context.ents:
    print(f"Entity Text: {ent.text}\tLabel: {ent.label_}")
```
